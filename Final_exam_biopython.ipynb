```python
from Bio.Blast import NCBIWWW, NCBIXML
from Bio import SeqIO

#Open the file
f = r"D:\dna2.fasta"

try:
    file=open(f)
except IOError:
    print("FILE NOT FOUND!!")
    
    
```

**Second Step:**

Read archive and build a dicctionary with  all the sequence and name them.

```python
#Dicctionary where will be storing our sequences
sequences={}

#reading and filling the dicctionary
for line in file:
    line= line.strip()
    if line.startswith('>'):
        words=line.split()
        name=words[0][1:]
        sequences[name]=''
    else:
        sequences[name]= sequences[name] + line
```

**Third Step:**

```python
#Longest and smallest sequences in file
#Max length sequence
longest= max(sequences,key=lambda n: len(sequences[n]))
print("\n Longest sequence in file:",longest,"--->", len(sequences[longest])," nt")

#Min length sequence
smallest= min(sequences,key=lambda n: len(sequences[n]))
print("\n Smallest sequence in file:",smallest,"---->", len(sequences[smallest])," nt")
```

```python
#Open reading frame
#Identify if a sequences has and OFR, meaning that it has a start codon and a finish codon

#ORF finder function
def orf_finder(dna,frame):
    start_codon_found= False
    start_codon= 'ATG'
    stop_codons=['TAA', 'TAG', 'TGA']
    orfs=[]

    for i in range (frame,len(dna)-2,3):
        codon= dna[i:i+3].upper()

        if not start_codon_found and codon == start_codon:
             start_codon_found= True   
             start_orf= i

        elif start_codon_found and codon in stop_codons:
            orfs.append((start_orf,i+3 - start_orf))
            start_codon_found = False
               
    return orfs


            
```

```python
## Usage of the function to find the longest ORF, its sequence identifier, the longest ORF contained in the sequence and the starting position of this Longest ORF in a frame number.
#General, from all sequences
longest_len=0
longest_id= None
longest_start= None
Longest_frame= None
Longest_seq= None


for name, dna in sequences.items():
    for frame in range(3):
        for start_orf,orf_len in orf_finder(dna,frame):
            if orf_len > longest_len:
                longest_len= orf_len
                longest_id= name
                longest_start= start_orf
                Longest_frame= frame+1
                Longest_seq= dna[start_orf:start_orf+ orf_len]

print("Name of the longest ORF:", longest_id)
print("Lenght of the longest ORF:", longest_len)
print("ORF starting position:", longest_start + 1 )
print("Frame used:", Longest_frame)
print("ORF Sequence:", Longest_seq)
```

```python
#Find Repeats on all sequences and identify how many times each repeat occurs in the file, and which is the most frequent repeat of a given length
#function to find repeats
def find_repeats(sequences,n):
    counts={}
    for dna in sequences.values():
        dna= dna.upper()
        for i in range(len(dna)- n + 1):
            sub= dna[i:i+n]
            counts[sub]= counts.get(sub,0) + 1
    return {sub: c for sub, c in counts.items() if c> 1}

```

```python
## Find wich is the mos frequent repeat of a given length

 #Asigned number of repeats to n, change if it needed to

repeats= find_repeats(sequences,n=7)

max_count= max(repeats.values())
most_frequent = [sub for sub, c in repeats.items() if c==max_count]

print("Different repeats of length",n,":", len(repeats))
print("Most frequent(s) repeat(s)", most_frequent)
print("Times repeated:",max_count)
print("How many different sequence occur max times",len(most_frequent))
```
